```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(r"C:\Users\hp\LOAN_DEFAULT_ANALYSIS\Cleaned_csv_file\Loan_default_clean.csv")
```

Total number of loans

```python
total_loans = len(df)

print("Total number of loans:", total_loans)
```

Total and average loan amount

```python
print("Total loan amount:",df["loan_amnt"].sum())
print("Average loan amount:",df["loan_amnt"].mean())
print("Median loan amount:",df["loan_amnt"].median())
```

Loan amount distribution

```python

plt.figure(figsize=(10, 6))

plt.hist(
    df["loan_amnt"].dropna(),
    bins=30
)

plt.title("Distribution of Loan Amount")
plt.xlabel("Loan Amount")
plt.ylabel("Number of Loans")

plt.show()
```

Interest rate distribution

```python
plt.figure(figsize=(10, 6))

plt.hist(
    df["int_rate"].dropna(),
    bins=30
)

plt.title("Distribution of Interest Rates")
plt.xlabel("Interest Rate (%)")
plt.ylabel("Number of Loans")

plt.show()
```

Loan grade distribution

```python
grade_counts = df["grade"].value_counts().sort_index()

plt.figure(figsize=(10, 6))

plt.bar(
    grade_counts.index,
    grade_counts.values
)

plt.title("Loan Distribution by Grade")
plt.xlabel("Loan Grade")
plt.ylabel("Number of Loans")

plt.show()
```

Loan purpose distribution

```python
purpose_counts = df["purpose"].value_counts()

plt.figure(figsize=(12, 6))

x = range(len(purpose_counts))

plt.bar(
    x,
    purpose_counts.values
)

plt.title("Loan Distribution by Purpose")
plt.xlabel("Loan Purpose")
plt.ylabel("Number of Loans")

plt.xticks(
    x,
    purpose_counts.index,
    rotation=45,
    ha="right"
)

plt.show()
```
