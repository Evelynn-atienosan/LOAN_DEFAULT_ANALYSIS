```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(r"C:\Users\hp\LOAN_DEFAULT_ANALYSIS\Cleaned_csv_file\Loan_default_clean.csv")
```

Annual income distribution

```python
plt.figure(figsize=(10, 6))

plt.boxplot(
    df["annual_inc"].dropna()
)

plt.title("Distribution of Borrower Annual Income")
plt.ylabel("Annual Income")

plt.show()
```

Median annual income

```python
print(
    "Median annual income:",
    df["annual_inc"].median()
)
```

Employment length

```python
emp_counts = df["emp_length"].value_counts()

plt.figure(figsize=(12, 6))

plt.bar(
    emp_counts.index.astype(str),
    emp_counts.values
)

plt.title("Borrowers by Employment Length")
plt.xlabel("Employment Length")
plt.ylabel("Number of Borrowers")

plt.xticks(rotation=45)

plt.show()
```

Home ownership

```python
home_counts = df["home_ownership"].value_counts()

plt.figure(figsize=(10, 6))

bars = plt.bar(
    home_counts.index,
    home_counts.values
)

for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        str(int(bar.get_height())),
        ha="center",
        va="bottom"
    )

plt.title("Loans by Home Ownership")
plt.xlabel("Home Ownership")
plt.ylabel("Number of Loans")

plt.show()
```

Verification status

```python
verification_counts = df["verification_status"].value_counts()

plt.figure(figsize=(10, 6))

plt.bar(
    verification_counts.index,
    verification_counts.values
)

plt.title("Loans by Verification Status")
plt.xlabel("Verification Status")
plt.ylabel("Number of Loans")

plt.xticks(rotation=30)

plt.show()
```
