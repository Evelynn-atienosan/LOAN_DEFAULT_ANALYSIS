Checking and removing duplicates for duplicated rows

```python
Loan_default.duplicated().sum()
```

```python
missing_values = Loan_default.isnull().sum().sort_values(ascending = False)
```

```python
missing_percentage = (Loan_default.isnull().sum()/len(Loan_default))* 100

missing_percentage
```

```python
missing_data = pd.DataFrame({
    "Missing Values": Loan_default.isnull().sum(),
    "Percentage": missing_percentage
}).sort_values(by = "Percentage", ascending = False)

missing_data
```

Removing comlumns with more than 80% missing values

```python
threshhold = 60

columns_to_drop =missing_data[missing_data["Percentage"] > threshhold].index

columns_to_drop
```

```python
Loan_default['is_joint_application'] = Loan_default['annual_inc_joint'].notna().astype(int)
```

```python
Loan_default_new = Loan_default.drop(columns = columns_to_drop )
```

```python
import numpy as np

Loan_default_new["default_status"] = np.select(
    [
        Loan_default_new["loan_status"].isin([
            "Charged Off",
            "Default",
            "Does not meet the credit policy. Status:Charged Off"
        ]),
        
        Loan_default_new["loan_status"].isin([
            "Fully Paid",
            "Does not meet the credit policy. Status:Fully Paid"
        ])
    ],
    [
        "Defaulted",
        "Did Not Default"
    ],
    default="Outcome Unknown"
)
```
